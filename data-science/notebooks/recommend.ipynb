```python
df['cast'] = df.cast.apply(lambda x: x.split(",")[:3] if pd.notna(x) else x)
```

```python
df['director'] = df.director.apply(lambda x: x.split(",")[0] if pd.notna(x) else x)
```

```python
df['cast'] = df.cast.fillna("")
df['director'] = df.director.fillna("")
```

```python
def collapse(x):
    l = []
    for i in x:
        l.append(i.replace(" ", ""))

    return l

```

```python
df['cast'] = df['cast'].apply(collapse)
```

```python
df['director'] = df['director'].apply(lambda x: x.replace(" ", ""))
```

```python
df['genres'] = df.genres.fillna("")
df['keywords'] = df.keywords.fillna("")
```

```python
df['genres'] = df.genres.apply(lambda x: x.replace(" ", ""))
```

```python
df.isna().sum()
```

```python


df = df[df.title.notnull()]
```

```python
df['overview'] = df.overview.apply(lambda x: x.split())
df['genres'] = df.genres.apply(lambda x: x.split(','))
```

```python
df['keywords'] = df.keywords.str.split("|").apply(collapse)
```

```python
df["genres"].apply(lambda x: x if len(x) < 2 else '')
```

```python
df['genres'] = df.genres.apply(lambda x: "" if x == [''] else x)
df['keywords'] = df.keywords.apply(lambda x: "" if x == [''] else x)
df['cast'] = df.cast.apply(lambda x: "" if x == [] else x)
```
